```python
import random
```

```python
data = []
```

```python
for i in range(500):
  cgpa = random.uniform(0,10)
  iq = random.randint(70,130)
  if cgpa >= 6 and iq > 100:
    result = 1
  else:
    result = 0
  data.append([cgpa,iq,result])
```

```python
print(type(data))
```

```python
df = pd.DataFrame(
    data,
    columns = ['cgpa','iq','result']
)
```

```python
# print(type(df))
df
```

```python
sns.scatterplot(
    data = df,
    x = 'cgpa',
    y = 'iq',
    hue = 'result'
)
```

```python
df.describe()
```

```python
X = df.iloc[:,0:2]
y = df.iloc[:,-1]
```

```python
X
```

```python
from sklearn.linear_model import Perceptron
```

```python
p = Perceptron()
```

```python
p.fit(X,y)
```

```python
p.coef_
```

```python
p.intercept_
```

```python
from mlxtend.plotting import plot_decision_regions
```

```python
plot_decision_regions(X.values,y.values,clf=p,legend=2)
```
